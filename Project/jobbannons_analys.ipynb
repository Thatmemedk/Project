## Testa JobTech API

Vi använder JobTechs API för att hämta jobbannonser.

API-anropet görs med biblioteket `requests`.

```python
url = "https://jobsearch.api.jobtechdev.se/search"

params = {
    "q": "AI utvecklare",
    "limit": 5
}

response = requests.get(url, params=params, timeout=10)

print("Statuskod:", response.status_code)
```

```python
if response.status_code == 200:
    print("API-anropet lyckades!")
else:
    print("Något gick fel.")
    print("Statuskod:", response.status_code)
```

```python
data = response.json()

print(type(data))
```

```python
print(data.keys())
```

```python
jobs = data.get("hits", [])

print("Antal jobbannonser:", len(jobs))
```

```python
if jobs:
    first_job = jobs[0]
    print("Första jobbannonsen:")
    print(json.dumps(first_job, indent=2, ensure_ascii=False))
else:
    print("Inga jobbannonser hittades.")
```
